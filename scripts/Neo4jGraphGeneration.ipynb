```python
from neo4j import GraphDatabase
import os,glob
from pathlib import Path
import json
```

```python
driver=GraphDatabase.driver(uri = "neo4j://127.0.0.1:7687",auth=("neo4j","Reasearch@2025"))
# session=driver.session()
```

```python
directory = Path.cwd().parent / "JSON-output"
# directory = Path("D:/Masters/Research Project/Project/output").expanduser().resolve()
pattern = os.path.join(directory, "*.json")

files = glob.glob(pattern)

print(files)
```

```python
query = """
WITH $doc AS d
MERGE (seed:Paper:Seed {id: coalesce(d.DOI, d.PaperTitle)})
  ON CREATE SET
    seed.PaperTitle      = d.PaperTitle,
    seed.PublicationYear = d.PublicationYear,
    seed.DOI             = d.DOI,
    seed.Authors         = d.Authors,
    seed.Abstract        = d.Abstract,
    seed.SummaryAbstract = d.SummaryAbstract

WITH seed, d
UNWIND coalesce(d.References, []) AS r
WITH seed, r
WHERE r.DOI IS NOT NULL AND trim(r.DOI) <> ""
  AND r.PaperTitle IS NOT NULL AND trim(r.PaperTitle) <> ""

MERGE (refPaper:Paper {id: coalesce(r.DOI, r.PaperTitle)})
  ON CREATE SET
    refPaper.PaperTitle      = r.PaperTitle,
    refPaper.PublicationYear = r.PublicationYear,
    refPaper.DOI             = r.DOI,
    refPaper.Authors         = r.Authors,
    refPaper.Editors         = r.Editors,
    refPaper.Publisher       = r.Publisher,
    refPaper.Publication     = r.Publication,
    refPaper.SemanticInformation = r.SemanticInformation

MERGE (seed)-[c:CITES]->(refPaper)
  ON CREATE SET
    c.SemanticInformation = r.SemanticInformation,
    c.CitedPublication    = r.Publication,
    c.CitedPublisher      = r.Publisher,
    c.CitedYear           = r.PublicationYear


WITH seed, refPaper, r
UNWIND coalesce(r.References, []) AS r2
WITH seed, refPaper, r2
WHERE r2.DOI IS NOT NULL AND trim(r2.DOI) <> ""
  AND r2.PaperTitle IS NOT NULL AND trim(r2.PaperTitle) <> ""

MERGE (ref2Paper:Paper {id: coalesce(r2.DOI, r2.PaperTitle)})
  ON CREATE SET
    ref2Paper.PaperTitle      = r2.PaperTitle,
    ref2Paper.PublicationYear = r2.PublicationYear,
    ref2Paper.DOI             = r2.DOI,
    ref2Paper.Authors         = r2.Authors,
    ref2Paper.Editors         = r2.Editors,
    ref2Paper.Publisher       = r2.Publisher,
    ref2Paper.Publication     = r2.Publication,
    ref2Paper.SemanticInformation = r2.SemanticInformation

MERGE (refPaper)-[cc:CITES]->(ref2Paper)
  ON CREATE SET
    cc.SemanticInformation = r2.SemanticInformation,
    cc.CitedPublication    = r2.Publication,
    cc.CitedPublisher      = r2.Publisher,
    cc.CitedYear           = r2.PublicationYear


WITH seed, refPaper, ref2Paper, r2
UNWIND coalesce(r2.References, []) AS r3
WITH seed, refPaper, ref2Paper, r3
WHERE r3.DOI IS NOT NULL AND trim(r3.DOI) <> ""
  AND r3.PaperTitle IS NOT NULL AND trim(r3.PaperTitle) <> ""

MERGE (ref3Paper:Paper {id: coalesce(r3.DOI, r3.PaperTitle)})
  ON CREATE SET
    ref3Paper.PaperTitle      = r3.PaperTitle,
    ref3Paper.PublicationYear = r3.PublicationYear,
    ref3Paper.DOI             = r3.DOI,
    ref3Paper.Authors         = r3.Authors,
    ref3Paper.Editors         = r3.Editors,
    ref3Paper.Publisher       = r3.Publisher,
    ref3Paper.Publication     = r3.Publication
    

MERGE (ref2Paper)-[ccc:CITES]->(ref3Paper)
  ON CREATE SET
    ccc.CitedPublication    = r3.Publication,
    ccc.CitedPublisher      = r3.Publisher,
    ccc.CitedYear           = r3.PublicationYear;
"""
```

```python
with driver.session(database="researchpapers") as session:
    for path in files:
        with open(path, "r", encoding="utf-8") as f:
            doc = json.load(f)
        session.run(query2, {"doc": doc})

driver.close()
```
